```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout
from tensorflow.keras.models import Model
import os

# ==========================================
# 1. SETUP PATHS (Apna Path Yahan Dalein)
# ==========================================
# Note: Path mein forward slash (/) use karein
dataset_path = "C:/Users/Front Man/OneDrive/Desktop/Water_Dataset_Final/images" 
# ==========================================

print(f"Checking dataset at: {dataset_path}")

# 2. ADVANCED DATA AUGMENTATION
# Images ko ghuma kar aur zoom kar ke AI ko sikhayega
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest',
    validation_split=0.2
)

print("Loading images...")

# Training Data
train_generator = train_datagen.flow_from_directory(
    dataset_path,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    subset='training'
)

# Validation Data
validation_generator = train_datagen.flow_from_directory(
    dataset_path,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    subset='validation'
)

print(f"Classes found: {train_generator.class_indices}")

# 3. BUILD MODEL (MobileNetV2)
base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
base_model.trainable = False 

# --- YAHAN CORRECTION KI HAI ---
x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(128, activation='relu')(x)
x = Dropout(0.5)(x)  # <--- Yahan (x) add kiya hai, ab error nahi ayega
predictions = Dense(3, activation='softmax')(x) 

model = Model(inputs=base_model.input, outputs=predictions)

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# 4. TRAIN
print("Starting Advanced Training... (Wait for 10-15 mins)")
history = model.fit(
    train_generator,
    epochs=15, 
    validation_data=validation_generator
)

# 5. CONVERT TO TFLITE
print("Converting model...")
converter = tf.lite.TFLiteConverter.from_keras_model(model)
tflite_model = converter.convert()

with open('water_quality_model.tflite', 'wb') as f:
    f.write(tflite_model)

print("SUCCESS! New 'water_quality_model.tflite' created.")
```
